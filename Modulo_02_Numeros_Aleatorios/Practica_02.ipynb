Curso de Simulación Estocástica — UJED


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(42)
u = rng.uniform(size=10000)
print("Media:", u.mean(), "Varianza:", u.var(ddof=1))
print("Prueba KS:", stats.kstest(u, "uniform"))
plt.hist(u, bins=30, density=True)
plt.xlabel("Valor")
plt.ylabel("Densidad")
plt.show()
```
