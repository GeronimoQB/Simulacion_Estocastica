```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
rng = np.random.default_rng(42)
x = rng.normal(loc=100, scale=15, size=10000)
print("Media muestral:", x.mean())
print("Desviación estándar muestral:", x.std(ddof=1))
t = np.linspace(40, 160, 400)
plt.hist(x, bins=40, density=True, alpha=0.6)
plt.plot(t, norm.pdf(t, 100, 15))
plt.xlabel("Valor")
plt.ylabel("Densidad")
plt.show()
```
