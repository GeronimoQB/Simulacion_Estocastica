# Determinístico y estocástico
Curso de Simulación Estocástica — UJED


```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(42)
x = np.arange(1, 101)
deterministico = 2 * x + 5
estocastico = deterministico + rng.normal(0, 12, len(x))
plt.plot(x, deterministico, label="Determinístico")
plt.scatter(x, estocastico, s=12, label="Estocástico")
plt.xlabel("Tiempo")
plt.ylabel("Respuesta")
plt.legend()
plt.show()
```
