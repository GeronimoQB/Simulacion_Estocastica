# Proceso de Poisson
Curso de Simulación Estocástica — UJED


```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(42)
tasa = 3.0
tiempo = np.linspace(0, 10, 501)
for i in range(5):
    eventos = rng.poisson(tasa * np.diff(tiempo))
    trayectoria = np.r_[0, np.cumsum(eventos)]
    plt.step(tiempo, trayectoria, where="post", label=f"Trayectoria {i+1}")
plt.xlabel("Tiempo")
plt.ylabel("Eventos acumulados")
plt.legend()
plt.show()
```
