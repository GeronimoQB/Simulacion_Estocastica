# Estimación de pi por Monte Carlo
Curso de Simulación Estocástica — UJED


```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(42)
n = 100000
puntos = rng.uniform(-1, 1, size=(n, 2))
dentro = (puntos**2).sum(axis=1) <= 1
estimacion = 4 * np.cumsum(dentro) / np.arange(1, n + 1)
print("Estimación de pi:", estimacion[-1])
print("Error absoluto:", abs(estimacion[-1] - np.pi))
plt.plot(estimacion)
plt.axhline(np.pi, linestyle="--")
plt.xlabel("Número de simulaciones")
plt.ylabel("Estimación de pi")
plt.show()
```
